```python
query = """
select * from globaltemperatures
"""
df = pd.read_sql(query, engine)
df['dt'] = pd.to_datetime(df['dt'])
df['year'] = df['dt'].dt.year

Avg_temp_yearly = (df.groupby('year')['Land_avg_temp']
                   .mean()
                   .reset_index())
display(Avg_temp_yearly)
```

```python
plt.figure(figsize=(12, 6))
plt.plot(
    Avg_temp_yearly['year'],
    Avg_temp_yearly['Land_avg_temp'])
plt.xlabel('Year')
plt.ylabel('Average land temperature in Celsius')
plt.title('Average land temperature by year')
plt.legend()
plt.grid(True)

plt.show()
```

```python
query = """
select * from globaltemperatures
"""
df = pd.read_sql(query, engine)
df['dt'] = pd.to_datetime(df['dt'])
df['century'] = ((df['dt'].dt.year - 1) // 100) + 1

Avg_temp_century = (df.groupby('century')['Land_avg_temp']
                   .mean()
                   .reset_index())
display(Avg_temp_century)
```

```python
plt.figure(figsize=(12, 6))
plt.plot(
    Avg_temp_century['century'],
    Avg_temp_century['Land_avg_temp'])
plt.xlabel('Century')
plt.ylabel('Average land temperature in Celsius')
plt.title('Average land temperature by century')
plt.xticks(Avg_temp_century['century'])
plt.grid()
plt.show()
```

```python
query = """
select * from temp_by_country
where country = 'Niger'
"""
query2 = """
select * from temp_by_country
where country = 'Greenland'
"""
df = pd.read_sql(query, engine)
df['dt'] = pd.to_datetime(df['dt'])
df['year'] = df['dt'].dt.year

Avg_temp_niger = (df.groupby(['country', 'year'])['avg_temp']
                   .mean()
                   .reset_index())
                   
df2 = pd.read_sql(query2, engine)
df2['dt'] = pd.to_datetime(df2['dt'])
df2['year'] = df2['dt'].dt.year

Avg_temp_greenland = (df2.groupby(['country', 'year'])['avg_temp']
                   .mean()
                   .reset_index())
display(Avg_temp_niger, Avg_temp_greenland)
```

```python
plt.figure(figsize=(12, 6))
# data
x = Avg_temp_niger['year']
y = Avg_temp_niger['avg_temp']
x2 = Avg_temp_greenland['year']
y2 = Avg_temp_greenland['avg_temp']
#plot
fig, ax = plt.subplots()
ax.plot(x, y, color = "Orange", label = "Niger")
ax.plot(x2, y2, color = "Blue", label = 'Greenland')
plt.xlabel('Year')
plt.ylabel('Average land temperature in Celsius')
plt.title('Average temp coldest vs hottest country')
plt.legend()
plt.grid()
plt.show()
```

```python
query = """
select * from temp_by_city
where avg_temp is not null
"""
df = pd.read_sql(query, engine)
df['dt'] = pd.to_datetime(df['dt'])
df['year'] = df['dt'].dt.year

display(df.describe())
```

```python
# prompt to select the 3 hottest city (no duplicate countries)
query = """
select * from temp_by_city
where avg_temp is not null and city = "Jibuti"
"""

df = pd.read_sql(query, engine)
df['dt'] = pd.to_datetime(df['dt'])
df['year'] = df['dt'].dt.year

Avg_temp_jibuti = (df.groupby(['city', 'year'])['avg_temp']
                   .mean()
                   .reset_index())

query2 = """
select * from temp_by_city
where avg_temp is not null and city = "Khartoum"
"""

df = pd.read_sql(query2, engine)
df['dt'] = pd.to_datetime(df['dt'])
df['year'] = df['dt'].dt.year

Avg_temp_Khartoum = (df.groupby(['city', 'year'])['avg_temp']
                   .mean()
                   .reset_index())

query3 = """
select * from temp_by_city
where avg_temp is not null and city = "Niamey"
"""

df = pd.read_sql(query3, engine)
df['dt'] = pd.to_datetime(df['dt'])
df['year'] = df['dt'].dt.year

Avg_temp_Niamey = (df.groupby(['city', 'year'])['avg_temp']
                   .mean()
                   .reset_index())
display(Avg_temp_jibuti, Avg_temp_Niamey, Avg_temp_Khartoum)
```

```python
#prompt to plot the 3 hottest cities from before
plt.figure(figsize=(16, 9))
# data
x = Avg_temp_jibuti['year']
y = Avg_temp_jibuti['avg_temp']
x2 = Avg_temp_Niamey['year']
y2 = Avg_temp_Niamey['avg_temp']
x3 = Avg_temp_Khartoum['year']
y3 = Avg_temp_Khartoum['avg_temp']
#plot
fig, ax = plt.subplots()
ax.plot(x, y, color = "Orange", label = "Jibuti (Djibouti)")
ax.plot(x2, y2, color = "Yellow", label = 'Niamey (Niger)')
ax.plot(x3, y3, color = "Brown", label = 'Khartoum (Sudan)')
plt.xlabel('Year')
plt.ylabel('Average temperature in Celsius')
plt.title('Average hottest cities')
plt.legend()
plt.grid()
plt.show()
```

```python
# prompt to select the 3 coldest cities (no duplicate countries)
query = """
select * from temp_by_city
where avg_temp is not null and city = "Norilsk"
"""

df = pd.read_sql(query, engine)
df['dt'] = pd.to_datetime(df['dt'])
df['year'] = df['dt'].dt.year

Avg_temp_Norilsk = (df.groupby(['city', 'year'])['avg_temp']
                   .mean()
                   .reset_index())

query2 = """
select * from temp_by_city
where avg_temp is not null and city = "Ulaanbaatar"
"""

df = pd.read_sql(query2, engine)
df['dt'] = pd.to_datetime(df['dt'])
df['year'] = df['dt'].dt.year

Avg_temp_Ulaanbaatar = (df.groupby(['city', 'year'])['avg_temp']
                   .mean()
                   .reset_index())

query3 = """
select * from temp_by_city
where avg_temp is not null and city = "Yakeshi"
"""

df = pd.read_sql(query3, engine)
df['dt'] = pd.to_datetime(df['dt'])
df['year'] = df['dt'].dt.year

Avg_temp_Yakeshi = (df.groupby(['city', 'year'])['avg_temp']
                   .mean()
                   .reset_index())
display(Avg_temp_Norilsk, Avg_temp_Ulaanbaatar, Avg_temp_Yakeshi)
```

```python
#prompt to plot the 3 coldest cities from before
plt.figure(figsize=(16, 9))
# data
xc = Avg_temp_Norilsk['year']
yc = Avg_temp_Norilsk['avg_temp']
xc2 = Avg_temp_Ulaanbaatar['year']
yc2 = Avg_temp_Ulaanbaatar['avg_temp']
xc3 = Avg_temp_Yakeshi['year']
yc3 = Avg_temp_Yakeshi['avg_temp']
#plot
fig, ax = plt.subplots()
ax.plot(xc, yc, color = "Blue", label = "Norilsk (Russia)")
ax.plot(xc2, yc2, color = "Cyan", label = 'Ulaanbaatar(Mongolia)')
ax.plot(xc3, yc3, color = "Purple", label = 'Yakeshi(China)')
plt.xlabel('Year')
plt.ylabel('Average temperature in Celsius')
plt.title('Average coldest cities')
plt.legend()
plt.grid()
plt.show()
```

```python
#Plotting hottest vs coldest cities
plt.figure(figsize=(16, 9))

xc = Avg_temp_Norilsk['year']
yc = Avg_temp_Norilsk['avg_temp']
x = Avg_temp_jibuti['year']
y = Avg_temp_jibuti['avg_temp']

fig, ax = plt.subplots()
ax.plot(x, y, color = 'Red', label = "Jibuti (Djibouti)")
ax.plot(xc, yc, color = "Blue", label = "Norilsk (Russia)")
plt.xlabel('Year')
plt.ylabel('Temperature in celsius')
plt.title("Hottest vs coldest city")
plt.legend()
plt.grid()
plt.show()
```
