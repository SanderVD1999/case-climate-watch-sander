```python
query = """
select dt, land_and_oceanavgtemp, land_avg_temp from globaltemperatures
where land_and_oceanavgtemp != 0 and land_avg_temp != 0
"""
df = pd.read_sql(query, engine)
df['dt'] = pd.to_datetime(df['dt'])
df['year'] = df['dt'].dt.year

temp_landsea = (df.groupby(['year'])[['land_avg_temp', 'land_and_oceanavgtemp']]
                   .mean()
                   .reset_index())
display(temp_landsea)
```

```python
#Plotting land temp and land and ocean temp combined
fig, ax = plt.subplots(figsize=(16, 9))

x = temp_landsea['year']
y = temp_landsea['land_avg_temp']
y2 = temp_landsea['land_and_oceanavgtemp']

ax.plot(x, y, color = 'Red', label = "Temperature on land")
ax.plot(x, y2, color = "Blue", label = "Temperature land and sea")
plt.xlabel('Year')
plt.ylabel('Temperature in celsius')
plt.title("Land temperature + combined with ocean temperature")
plt.legend()
plt.grid()
plt.show()
```

```python
query = """
select dt, land_avg_temp, land_avg_temp_uncertainty, land_and_oceanavgtemp, land_and_oceanavgtemp_unc from globaltemperatures
where land_and_oceanavgtemp_unc is not null and land_avg_temp_uncertainty is not null 
"""
df = pd.read_sql(query, engine)
df['dt'] = pd.to_datetime(df['dt'])
df['year'] = df['dt'].dt.year

temp_unc = (df.groupby(['year'])[['land_avg_temp_uncertainty', 'land_and_oceanavgtemp_unc']]
                   .mean()
                   .reset_index())
display(temp_unc)
```

```python
fig, ax = plt.subplots(figsize=(16, 9))

x = temp_unc['year']
y = temp_unc['land_avg_temp_uncertainty']
y2 = temp_unc['land_and_oceanavgtemp_unc']

ax.plot(x, y, color = 'Brown', label = "uncertainty Temperature on land")
ax.plot(x, y2, color = 'Blue', label = "Uncertainty temperature on land and ocean")
plt.xlabel('Year')
plt.ylabel('Temperature in celsius')
plt.title("Temperature uncertainty by the years")
plt.legend()
plt.grid()
plt.show()
```

```python
import plotly.express as px

query = """
select latitude, longitude, avg_temp, city, country from temp_by_city
where avg_temp is not null and avg_temp != 0
"""
df = pd.read_sql(query, engine)
df['latitude'] = (
    df['latitude'].str[:-1].astype(float) *
    df['latitude'].str[-1].map({'N': 1, 'S': -1}))

df['longitude'] = (
    df['longitude'].str[:-1].astype(float) *
    df['longitude'].str[-1].map({'E': 1, 'W': -1}))

temp_map = (df.groupby(['city']).agg(
                                    avg_temp=('avg_temp', 'mean'),
                                    latitude=('latitude', 'first'),
                                    longitude=('longitude', 'first'))    
                                .reset_index())

display(temp_map)
```

```python
fig = px.scatter_geo(temp_map,
                    lat='latitude',
                    lon='longitude',
                    color='avg_temp',
                    hover_name="city",
                    )
fig.show()
```
